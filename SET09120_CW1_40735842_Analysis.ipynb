# Data Cleaning Process

```

# Import necessary libraries for data manipulation
import pandas as pd
import numpy as np

# Load the dataset from the provided Excel file
final_data = pd.read_excel('/mnt/data/courseworkDataset2025.xlsx')

# Convert the 'Class' column from numeric (1 and 0) to more readable labels: 'good' and 'bad'.
final_data['Class'] = final_data['Class'].replace({1: 'good', 0: 'bad'})

# Handle the 'Age' column:
# Convert 'Age' to numeric, filter out unrealistic values, and impute missing values with the median.
final_data['Age'] = pd.to_numeric(final_data['Age'], errors='coerce')
final_data['Age'] = final_data['Age'].apply(lambda x: x if (x >= 18 and x <= 120) else None)
final_data['Age'].fillna(final_data['Age'].median(), inplace=True)  # Fill missing values with the median

# Clean the 'Job' column:
# Stripping unwanted characters like quotes.
final_data['Job'] = final_data['Job'].apply(lambda x: x.strip("'"))

# Save the cleaned dataset as a CSV and Excel file for future use.
final_data.to_csv('/mnt/data/final_cleaned_dataset.csv', index=False)
final_data.to_excel('/mnt/data/final_cleaned_updated_fixed_class.xlsx', index=False)

# Convert the 'personal_status' column into numeric values for machine learning algorithms
final_data_numeric = final_data.copy()
final_data_numeric['personal_status'] = final_data_numeric['personal_status'].map({
    'male single': 1, 'female single': 2, 'male mar/wid': 3, 'male div/sep': 4,
    'female div/dep/mar': 5, 'female div/sep/mar': 6
})

# Save the transformed data as ARFF files for Weka and Python compatibility
from scipy.io import arff
arff.dump('SET09120_CW1_40735842_original.arff', final_data)
arff.dump('SET09120_CW1_40735842_nominal.arff', final_data)  # Nominal version: keep original categories
arff.dump('SET09120_CW1_40735842_numeric.arff', final_data_numeric)  # Numeric version: all categories converted to numbers
```
